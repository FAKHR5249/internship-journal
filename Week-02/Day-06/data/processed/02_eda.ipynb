# Exploratory Data Analysis (EDA)

## Objectives

- Understand target distribution
- Explore time-based patterns
- Analyze geographic features
- Study correlations
- Identify outliers

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

plt.style.use("ggplot")v
```

```python
df = pd.read_parquet("../data/processed/featured_taxi_data.parquet")

print("Dataset Loaded")
print(df.shape)
```

## Trip Duration Distribution

```python
plt.figure(figsize=(10,5))

plt.hist(df["trip_duration_minutes"], bins=60)

plt.title("Trip Duration Distribution")

plt.xlabel("Minutes")

plt.ylabel("Frequency")

plt.show()
```

## Log Transformed Target

```python
log_target = np.log1p(df["trip_duration_minutes"])

plt.figure(figsize=(10,5))

plt.hist(log_target, bins=60)

plt.title("Log Trip Duration")

plt.show()
```

## Average Trip Duration by Pickup Hour

```python
hourly = df.groupby("pickup_hour")["trip_duration_minutes"].mean()

plt.figure(figsize=(10,5))

hourly.plot(kind="bar")

plt.title("Average Trip Duration by Hour")

plt.ylabel("Minutes")

plt.show()
```

## Average Trip Duration by Day of Week

```python
weekday = df.groupby("day_of_week")["trip_duration_minutes"].mean()

plt.figure(figsize=(10,5))

weekday.plot(kind="bar")

plt.title("Average Trip Duration by Day")

plt.ylabel("Minutes")

plt.show()
```

## Average Trip Duration by Month

```python
monthly = df.groupby("month")["trip_duration_minutes"].mean()

plt.figure(figsize=(8,4))

monthly.plot(marker="o")

plt.title("Monthly Average Trip Duration")

plt.ylabel("Minutes")

plt.show()
```

## Passenger Count Distribution

```python
plt.figure(figsize=(8,4))

df["passenger_count"].value_counts().sort_index().plot(kind="bar")

plt.title("Passenger Count")

plt.show()
```

## Trip Distance Distribution

```python
plt.figure(figsize=(10,5))

plt.hist(df["trip_distance"], bins=50)

plt.title("Trip Distance")

plt.xlabel("Miles")

plt.show()
```

## Peak vs Off-Peak Trips

```python
peak = df.groupby("is_peak_hour")["trip_duration_minutes"].mean()

plt.figure(figsize=(6,4))

peak.plot(kind="bar")

plt.xticks([0,1],["Off Peak","Peak"])

plt.ylabel("Minutes")

plt.show()
```

## Weekend vs Weekday

```python
weekend = df.groupby("is_weekend")["trip_duration_minutes"].mean()

plt.figure(figsize=(6,4))

weekend.plot(kind="bar")

plt.xticks([0,1],["Weekday","Weekend"])

plt.ylabel("Minutes")

plt.show()
```

## Correlation Matrix

```python
corr = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(12,10))

plt.imshow(corr, cmap="coolwarm")

plt.colorbar()

plt.title("Correlation Matrix")

plt.show()
```

## Summary

- Trip duration is right-skewed.
- Log transformation reduces skewness.
- Peak hours generally have longer trips.
- Passenger count has limited influence.
- Distance is positively related to duration.
- Dataset is ready for regression modeling.
